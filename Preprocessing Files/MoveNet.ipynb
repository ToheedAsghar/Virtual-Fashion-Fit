```python
import os

def clean_folder_by_prefixes(folder_path, valid_prefixes):
    # Convert to set for fast lookup
    valid_prefixes = set(valid_prefixes)

    # List all files in the folder
    for filename in os.listdir(folder_path):
        file_path = os.path.join(folder_path, filename)
        if not os.path.isfile(file_path):
            continue

        # Get filename without extension
        name_wo_ext = os.path.splitext(filename)[0]

        # Check if it starts with any of the valid prefixes
        if not any(name_wo_ext.startswith(prefix) for prefix in valid_prefixes):
            try:
                os.remove(file_path)
                print(f"Deleted: {filename}")
            except Exception as e:
                print(f"Error deleting {filename}: {e}")
        else:
            print(f"Kept: {filename}")

# === Example usage in Google Colab ===
folder_path = "/content/image"  # Change to your folder in Colab

# Combine both lists and remove duplicates
valid_prefixes = [
    "00466_00", "00470_00", "01101_00", "01210_00", "01247_00", "01249_00",
    "01096_00", "01266_00"
]

clean_folder_by_prefixes(folder_path, valid_prefixes)
```

```python
# -*- coding: utf-8 -*-
import tensorflow as tf
import tensorflow_hub as hub
import numpy as np
import cv2
import json
import os
from tqdm import tqdm

def movenet_to_coco19_with_midhip(movenet_keypoints):
    """
    Converts MoveNet's 17 keypoints into 19 keypoints:
      - Insert a Neck (avg of shoulders)
      - Insert a MidHip (avg of hips)
    Final indexing (19 keypoints), adapted from an OpenPose-like format:
      0  : Nose
      1  : Neck
      2  : RShoulder
      3  : RElbow
      4  : RWrist
      5  : LShoulder
      6  : LElbow
      7  : LWrist
      8  : MidHip
      9  : RHip
      10 : RKnee
      11 : RAnkle
      12 : LHip
      13 : LKnee
      14 : LAnkle
      15 : REye
      16 : LEye
      17 : REar
      18 : LEar
    """

    # MoveNet indices
    NOSE = 0
    LEFT_EYE = 1
    RIGHT_EYE = 2
    LEFT_EAR = 3
    RIGHT_EAR = 4
    LEFT_SHOULDER = 5
    RIGHT_SHOULDER = 6
    LEFT_ELBOW = 7
    RIGHT_ELBOW = 8
    LEFT_WRIST = 9
    RIGHT_WRIST = 10
    LEFT_HIP = 11
    RIGHT_HIP = 12
    LEFT_KNEE = 13
    RIGHT_KNEE = 14
    LEFT_ANKLE = 15
    RIGHT_ANKLE = 16

    # Prepare an array of shape (19, 3)
    coco19 = np.zeros((19, 3), dtype=np.float32)

    # 0: Nose
    coco19[0] = movenet_keypoints[NOSE]

    # 1: Neck (avg of left & right shoulders)
    lsh = movenet_keypoints[LEFT_SHOULDER]
    rsh = movenet_keypoints[RIGHT_SHOULDER]
    neck_y = (lsh[0] + rsh[0]) / 2.0
    neck_x = (lsh[1] + rsh[1]) / 2.0
    neck_conf = (lsh[2] + rsh[2]) / 2.0
    coco19[1] = [neck_y, neck_x, neck_conf]

    # 2: RShoulder
    coco19[2] = movenet_keypoints[RIGHT_SHOULDER]
    # 3: RElbow
    coco19[3] = movenet_keypoints[RIGHT_ELBOW]
    # 4: RWrist
    coco19[4] = movenet_keypoints[RIGHT_WRIST]

    # 5: LShoulder
    coco19[5] = movenet_keypoints[LEFT_SHOULDER]
    # 6: LElbow
    coco19[6] = movenet_keypoints[LEFT_ELBOW]
    # 7: LWrist
    coco19[7] = movenet_keypoints[LEFT_WRIST]

    # 8: MidHip (avg of left & right hips)
    lhip = movenet_keypoints[LEFT_HIP]
    rhip = movenet_keypoints[RIGHT_HIP]
    midhip_y = (lhip[0] + rhip[0]) / 2.0
    midhip_x = (lhip[1] + rhip[1]) / 2.0
    midhip_conf = (lhip[2] + rhip[2]) / 2.0
    coco19[8] = [midhip_y, midhip_x, midhip_conf]

    # 9 : RHip
    coco19[9] = movenet_keypoints[RIGHT_HIP]
    # 10: RKnee
    coco19[10] = movenet_keypoints[RIGHT_KNEE]
    # 11: RAnkle
    coco19[11] = movenet_keypoints[RIGHT_ANKLE]

    # 12: LHip
    coco19[12] = movenet_keypoints[LEFT_HIP]
    # 13: LKnee
    coco19[13] = movenet_keypoints[LEFT_KNEE]
    # 14: LAnkle
    coco19[14] = movenet_keypoints[LEFT_ANKLE]

    # 15: REye
    coco19[15] = movenet_keypoints[RIGHT_EYE]
    # 16: LEye
    coco19[16] = movenet_keypoints[LEFT_EYE]
    # 17: REar
    coco19[17] = movenet_keypoints[RIGHT_EAR]
    # 18: LEar
    coco19[18] = movenet_keypoints[LEFT_EAR]

    return coco19

def render_keypoints_coco19(image, keypoints, output_path):
    """
    Renders the 19 keypoints (COCO + MidHip) on a black background.
    Index layout:
      8 = MidHip
      9 = RHip, 12 = LHip
      etc.
    """

    # Create a black canvas the same size as original
    black_bg = np.zeros_like(image)

    # Example BGR color palette for 19 indices
    # (You can adjust each color to your liking.)
    keypoint_colors = {
        0:  (85,   0, 255),    # Nose
        1:  ( 0,   0, 255),    # Neck
        2:  ( 0,  85, 255),    # RShoulder
        3:  ( 0, 170, 255),    # RElbow
        4:  ( 0, 255, 255),    # RWrist
        5:  ( 0, 255, 170),    # LShoulder
        6:  ( 0, 255,  85),    # LElbow
        7:  ( 0, 255,   0),    # LWrist
        8:  (85, 255,  85),    # MidHip
        9:  (85, 255,   0),    # RHip
        10: (170,255,   0),    # RKnee
        11: (255,255,   0),    # RAnkle
        12: (255,170,   0),    # LHip
        13: (255, 85,   0),    # LKnee
        14: (255,  0,   0),    # LAnkle
        15: (255,  0,  85),    # REye
        16: (255,  0, 170),    # LEye
        17: (255,  0, 255),    # REar
        18: (170,  0, 255),    # LEar
    }

    # Connections reflect MidHip usage
    # Typical OpenPose skeleton with MidHip is:
    connections = [
        (0, 1),    # Nose -> Neck
        (1, 2),    # Neck -> RShoulder
        (2, 3),    # RShoulder -> RElbow
        (3, 4),    # RElbow -> RWrist
        (1, 5),    # Neck -> LShoulder
        (5, 6),    # LShoulder -> LElbow
        (6, 7),    # LElbow -> LWrist

        (1, 8),    # Neck -> MidHip

        (8, 9),    # MidHip -> RHip
        (9,10),    # RHip -> RKnee
        (10,11),   # RKnee -> RAnkle

        (8,12),    # MidHip -> LHip
        (12,13),   # LHip -> LKnee
        (13,14),   # LKnee -> LAnkle

        (0,15),    # Nose -> REye
        (15,17),   # REye -> REar
        (0,16),    # Nose -> LEye
        (16,18)    # LEye -> LEar
    ]

    line_thickness = 5
    circle_radius = 8
    conf_threshold = 0.1

    # Draw each keypoint
    for i, (y, x, conf) in enumerate(keypoints):
        if conf < conf_threshold:
            continue
        color = keypoint_colors.get(i, (255, 255, 255))
        cv2.circle(black_bg, (int(x), int(y)), circle_radius, color, -1)

    # Draw connections
    for (start_idx, end_idx) in connections:
        y1, x1, c1 = keypoints[start_idx]
        y2, x2, c2 = keypoints[end_idx]
        if c1 > conf_threshold and c2 > conf_threshold:
            color = keypoint_colors.get(start_idx, (255, 255, 255))
            cv2.line(
                black_bg,
                (int(x1), int(y1)),
                (int(x2), int(y2)),
                color,
                line_thickness
            )

    # Save the result
    cv2.imwrite(output_path, black_bg)

def process_images(
    input_dir,
    output_json_dir,
    output_rendered_dir,
    model_url='https://tfhub.dev/google/movenet/singlepose/lightning/4'
):
    """
    1) Loads MoveNet (SinglePose)
    2) For each image, extracts 17 keypoints
    3) Inserts MidHip => total 19 keypoints
    4) Saves JSON with 19 * 3 = 57 values in pose_keypoints_2d
    5) Renders black-background skeleton with MidHip
    """

    print("Loading MoveNet from TF-Hub...")
    model = hub.load(model_url)
    print("MoveNet model loaded successfully.")

    os.makedirs(output_json_dir, exist_ok=True)
    os.makedirs(output_rendered_dir, exist_ok=True)

    # Gather valid image files
    image_files = [
        f for f in os.listdir(input_dir)
        if f.lower().endswith(('.png', '.jpg', '.jpeg', '.webp', '.bmp'))
    ]
    print(f"Found {len(image_files)} image(s) in {input_dir}.")

    for filename in tqdm(image_files, desc="Processing Images"):
        image_path = os.path.join(input_dir, filename)
        image_bgr = cv2.imread(image_path)
        if image_bgr is None:
            print(f"Warning: Could not read {image_path}, skipping.")
            continue

        # Convert BGR -> RGB
        image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

        # Resize w/ pad to 192x192 for MoveNet
        input_image = tf.image.resize_with_pad(
            np.expand_dims(image_rgb, axis=0),
            192,
            192
        )
        input_image = tf.cast(input_image, dtype=tf.int32)

        # Inference
        outputs = model.signatures['serving_default'](input_image)
        keypoints_17 = outputs['output_0'].numpy().reshape((17, 3))

        # Scale back to original size
        h, w, _ = image_bgr.shape
        keypoints_17[:, 0] *= h  # y
        keypoints_17[:, 1] *= w  # x

        # Convert to 19 keypoints with MidHip
        coco19_keypoints = movenet_to_coco19_with_midhip(keypoints_17)

        # Build array for JSON [x0, y0, c0, x1, y1, c1, ...]
        pose_keypoints_2d = []
        for (y, x, conf) in coco19_keypoints:
            pose_keypoints_2d.extend([float(x), float(y), float(conf)])

        # JSON structure
        keypoints_dict = {
            "version": 1.3,
            "people": [
                {
                    "person_id": [-1],
                    # now 19 * 3 = 57 float values
                    "pose_keypoints_2d": pose_keypoints_2d,
                    "face_keypoints_2d": [],
                    "hand_left_keypoints_2d": [],
                    "hand_right_keypoints_2d": [],
                    "pose_keypoints_3d": [],
                    "face_keypoints_3d": [],
                    "hand_left_keypoints_3d": [],
                    "hand_right_keypoints_3d": []
                }
            ]
        }

        # Save JSON
        json_filename = f"{os.path.splitext(filename)[0]}_keypoints.json"
        output_json_path = os.path.join(output_json_dir, json_filename)
        with open(output_json_path, 'w') as jf:
            json.dump(keypoints_dict, jf, indent=2)

        # Render skeleton with 19 keypoints
        rendered_filename = f"{os.path.splitext(filename)[0]}_rendered.png"
        rendered_path = os.path.join(output_rendered_dir, rendered_filename)

        render_keypoints_coco19(image_bgr, coco19_keypoints, rendered_path)

        print(f"Processed {filename} -> {json_filename} & {rendered_filename}")


    # Adjust these to your own directories:
input_dir = "/content/image"
output_json_dir = "/content/movenet_json"
output_rendered_dir = "/content/movenet_img"

process_images(input_dir, output_json_dir, output_rendered_dir)
```
