```python
import os

def clean_folder_by_prefixes(folder_path, valid_prefixes):
    # Convert to set for fast lookup
    valid_prefixes = set(valid_prefixes)

    # List all files in the folder
    for filename in os.listdir(folder_path):
        file_path = os.path.join(folder_path, filename)
        if not os.path.isfile(file_path):
            continue

        # Get filename without extension
        name_wo_ext = os.path.splitext(filename)[0]

        # Check if it starts with any of the valid prefixes
        if not any(name_wo_ext.startswith(prefix) for prefix in valid_prefixes):
            try:
                os.remove(file_path)
                print(f"Deleted: {filename}")
            except Exception as e:
                print(f"Error deleting {filename}: {e}")
        else:
            print(f"Kept: {filename}")

# === Example usage in Google Colab ===
folder_path = "/content/detectron2/projects/DensePose/image_path"  # Change to your folder path in Colab

valid_prefixes = [
    "02513_00", "02311_00", "01004_00", "00124_00", "02389_00", "02011_00", "02204_00", "02613_00", "03025_00",
    "00475_00", "00534_00", "03203_00", "02090_00", "02110_00", "00125_00", "02824_00", "00328_00", "00344_00",
    "00718_00", "01126_00", "02271_00", "00292_00", "02106_00", "03167_00", "01638_00", "00349_00", "01247_00",
    "01489_00", "01197_00", "02276_00", "00162_00", "02952_00", "00828_00", "00716_00", "02128_00", "02203_00",
    "01793_00", "02523_00", "01266_00", "01617_00", "00506_00", "03085_00", "02178_00", "02174_00", "03108_00",
    "00207_00", "01932_00", "00226_00", "02469_00", "01073_00", "01674_00", "00337_00", "03031_00", "00200_00",
    "02966_00", "00472_00", "02443_00", "02296_00", "00171_00", "00164_00", "02427_00", "03106_00", "01060_00",
    "03010_00", "02360_00", "02914_00", "01935_00", "02114_00", "00682_00", "00235_00", "00130_00", "00823_00",
    "02044_00", "02982_00", "01725_00", "03027_00", "01212_00", "01669_00", "02589_00", "00919_00", "03236_00",
    "02413_00", "01608_00", "00230_00", "02240_00", "00375_00", "02000_00", "02313_00", "00857_00", "03082_00",
    "02109_00", "03089_00", "02630_00", "00811_00", "03030_00", "00382_00", "00260_00", "02193_00", "01937_00",
    "01101_00", "02119_00", "01018_00", "02388_00", "01195_00", "00174_00", "00341_00", "02662_00", "01211_00",
    "01070_00", "00815_00", "01659_00", "03154_00", "02359_00", "02545_00", "00100_00", "00106_00", "00091_00",
    "02631_00", "03055_00", "01069_00", "02535_00", "03155_00", "01796_00", "02149_00", "02170_00", "03122_00",
    "02375_00", "00378_00", "01439_00", "03212_00", "00385_00", "02316_00", "02955_00", "02964_00", "01606_00",
    "02054_00", "00541_00", "01988_00", "03020_00", "03170_00", "02088_00", "02521_00", "02308_00", "00215_00",
    "02424_00", "02883_00", "00527_00", "01944_00", "03132_00", "03006_00", "02220_00", "01498_00", "02189_00",
    "02429_00", "03205_00", "00821_00", "00530_00", "00483_00", "02056_00", "01797_00", "03216_00", "01628_00",
    "02208_00", "00332_00", "02466_00", "02813_00", "02958_00", "02160_00", "01580_00", "01494_00", "02145_00",
    "00501_00", "01590_00", "02055_00", "01198_00", "02377_00", "01717_00", "02603_00", "02946_00", "01574_00",
    "02846_00", "00394_00", "03019_00", "01637_00", "02811_00", "02225_00", "02962_00", "01798_00", "00165_00",
    "00216_00", "00484_00", "01413_00", "01680_00", "02121_00", "01994_00", "03084_00", "03195_00", "00248_00",
    "02356_00", "01098_00", "01737_00", "00358_00", "02195_00", "01086_00", "02996_00", "00829_00", "01936_00",
    "01635_00", "02626_00", "03206_00", "00496_00", "00462_00", "02024_00", "00169_00", "00166_00", "00142_00",
    "00306_00", "02198_00", "02307_00", "00309_00", "02825_00", "00498_00", "00330_00", "02940_00", "02099_00",
    "00466_00", "00247_00", "03136_00", "02905_00", "02812_00", "01271_00", "03112_00", "02157_00", "02953_00"
]

clean_folder_by_prefixes(folder_path, valid_prefixes)
```
